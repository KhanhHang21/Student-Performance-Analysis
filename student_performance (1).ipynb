```python
# Plot percentage of readmissions across primary diagnosis groups
import matplotlib.pyplot as plt

def plot_stacked_bar_chart(df, categorical_variable_1, categorical_variable_2):
    """
    Plot a stacked bar chart of categorical_variable_2 by categorical_variable_1.

    Parameters:
    df (pd.DataFrame): DataFrame containing categorical_variable_1 and categorical_variable_2.
    categorical_variable_1 (str): The column name for the variable to group by on the x-axis.
    categorical_variable_2 (str): The column name for the variable to stack in the bars.
    """

    # Group and unstack the data to prepare for plotting
    stacked_counts = df.groupby([categorical_variable_1, categorical_variable_2]).size().unstack(fill_value=0)

    # Ensure correct order if there's a specific ordering of categories (optional)
    # stacked_counts = stacked_counts[seniority_order]  # Uncomment and customize if needed

    # Convert counts to percentages
    stacked_percentages = stacked_counts.div(stacked_counts.sum(axis=1), axis=0) * 100

    # Define a color palette (customize as needed)
    colors = ['#d9fbd3', '#a8e6a1', '#6ac36a', '#2e7d32', '#1b5e20'] # Modify based on your number of categories

    # Plot the stacked bar chart
    ax = stacked_percentages.plot(kind='bar', stacked=True, figsize=(11, 8), color=colors)

    # Add percentage labels on the bars, only if percentage is greater than or equal to 1%
    for container in ax.containers:
        labels = [f'{v:.1f}%' if v >= 1 else '' for v in container.datavalues]
        ax.bar_label(container, labels=labels, label_type='center', color='white')

    # Set titles and labels
    plt.xlabel(categorical_variable_1, fontsize=12)
    plt.ylabel("Percentage", fontsize=12)
    plt.title(f"Stacked Bar Chart of {categorical_variable_2} by {categorical_variable_1} (Percentage)", fontsize=14)
    plt.xticks(rotation=0)
    plt.legend(title=categorical_variable_2, bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(axis='y', linestyle='-', linewidth=0.7, which='major', color='grey')
    plt.grid(axis='x', visible=False)

    # Show the plot
    plt.tight_layout()
    
plot_stacked_bar_chart(df, 'ParentalSupport', 'ParentalEducation')
```

```python
from scipy import stats
from tabulate import tabulate

def chi_square_test_of_independence(df, categorical_variable_1, categorical_variable_2):
    """
    Perform a Chi-square test of independence/association between two categorical variables.

    Parameters:
    df (pd.DataFrame): DataFrame containing the categorical variables.
    categorical_variable_1 (str): The first categorical variable.
    categorical_variable_2 (str): The second categorical variable.

    Returns:
    None. Prints the Chi-square test results in a table format.
    """
    # Create a contingency table
    contingency_table = pd.crosstab(df[categorical_variable_1], df[categorical_variable_2])

    # Perform the Chi-square test
    chi2, p, dof, expected = stats.chi2_contingency(contingency_table)

    # Create a metrics table with tabulate
    headers = ['Metric', 'Value']
    table = [
        ['Chi-Square Statistic', f'{chi2:.4f}'],
        ['p-value', f'{p:.4f}'],
        ['Degrees of Freedom', dof],
    ]

    # Print the results in a neat format
    print(tabulate(table, headers, tablefmt="pretty"))

    # Optional: Return the contingency table for further inspection
    return contingency_table
```

### **Results from One-way ANOVA Test**

**F-statistic:** 1.81

**P-value:** 0.1245

**Interpretation:**

F-statistic: The F-statistic of 1.81 suggests that **there is some variability in GPA across the different Parental Education groups, but the difference is not very large.**

P-value: The P-value of 0.1245 is greater than 0.05, meaning that **we fail to reject the null hypothesis. This indicates that there is no statistically significant difference in the mean GPA across the different levels of Parental Education.**

Conclusion:
Based on the ANOVA test, **Parental Education does not seem to have a significant effect on GPA** in this dataset. The difference in mean GPA across the five Parental Education groups is not statistically significant.

```python
from scipy.stats import f_oneway

# Group GPA by Parental Education Groups
grouped_gpa = [df[df['ParentalEducation'] == group]['GPA'] for group in df['ParentalEducation'].unique()]

# Perform ANOVA test
f_stat, p_value = f_oneway(*grouped_gpa)

# Display the results
f_stat, p_value
```

### Findings:

1) **No Strong Relationship Between Parental Education and GPA:**

The plot suggests that there is no clear trend between Parental Education and GPA. The median GPA remains relatively consistent across all parental education levels, indicating that parental education may not be a strong predictor of student performance in this dataset.

2) **Similar Spread Across Groups:**

The spread of GPAs (IQR) is similar for all education levels, showing that the variation in GPA does not drastically differ based on parental education. This suggests that other factors besides parental education may play a more significant role in academic performance.

3) **Absence of Extreme Outliers:**

No extreme outliers were identified for any group, which could suggest that the dataset is generally well-behaved in terms of outliers for GPA.

### **Regression Analysis**
- Perform simple linear regression for **no. of absent days (dependent variable) vs GPA (independent variable)**
- Breakdown by student age/grade
- Obtain metrics: RMSE, R-squared, P-values for slope and intercept

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import statsmodels.api as sm
from tabulate import tabulate

def linear_regression(df, target, dependent_var, hue=None):
    """ Create a scatter plot with linear regression line and optional hue for color grouping."""

    # Set figsize
    plt.figure(figsize=(10, 7))

    # Scatter plot
    sns.scatterplot(
        data=df,
        x=dependent_var,
        y=target,
        hue=hue,
        palette='Set1',
        s=75,
        edgecolor='black',
        alpha=0.3
    )

    # Prepare data for regression
    X = df[dependent_var].values.reshape(-1, 1)
    y = df[target].values

    # Perform linear regression
    reg = LinearRegression()
    reg.fit(X, y)
    y_pred = reg.predict(X)

    # Plot regression line
    plt.plot(df[dependent_var], y_pred, color='red', linewidth=2,
             label=f'Best Fit Line: y = {reg.coef_[0]:.2f}x + {reg.intercept_:.2f}')

    # Metrics
    rmse = np.sqrt(mean_squared_error(y, y_pred))
    r2 = r2_score(y, y_pred)

    # Titles and labels
    plt.title(f'{target} vs {dependent_var}', fontsize=14)
    plt.xlabel(dependent_var, fontsize=12)
    plt.ylabel(target, fontsize=12)

    if hue:
        plt.legend(title=hue, loc='upper left', frameon=True, borderpad=1, fancybox=True)
    else:
        plt.legend(loc='upper left', frameon=True, borderpad=1, fancybox=True)

    plt.grid(True)
    plt.tight_layout()
    plt.show()

    # OLS regression using statsmodels
    X_sm = sm.add_constant(X)
    model = sm.OLS(y, X_sm)
    results = model.fit()

    # Extract regression results
    slope = reg.coef_[0]
    intercept = reg.intercept_
    p_values = results.pvalues
    conf_int = results.conf_int()

    # Display summary in table
    headers = ['Metric', 'Value']
    table = [
        ['Slope', f'{slope:.4f}'],
        ['Intercept', f'{intercept:.4f}'],
        ['Root Mean Squared Error', f'{rmse:.4f}'],
        ['R-squared', f'{r2:.4f}'],
        ['P-value (Intercept)', f'{p_values[0]:.4f}'],
        ['P-value (Slope)', f'{p_values[1]:.4f}'],
        ['95% CI (Intercept)', f'[{conf_int[0][0]:.4f}, {conf_int[0][1]:.4f}]'],
        ['95% CI (Slope)', f'[{conf_int[1][0]:.4f}, {conf_int[1][1]:.4f}]'],
    ]
```

```python
# Run linear regression model on entire dataset - no. of absent days (dependent variable) vs GPA (independent variable)
linear_regression(df, 'GPA', 'Absences')
```

### **Two-sample T-Test**
we will compare the mean GPA of two groups:

a) Students who received tutoring (coded as 1).

b) Students who did not receive tutoring (coded as 0).

The T-test will help **determine if there is a statistically significant difference in the mean GPA between these two groups.**

```python
from scipy.stats import ttest_ind

# Separate the data into two groups: Tutoring (1) vs. No Tutoring (0)
tutoring_gpa = df[df['Tutoring'] == 'Yes']['GPA']
no_tutoring_gpa = df[df['Tutoring'] == 'No']['GPA']

# Perform T-test
t_stat, p_value = ttest_ind(tutoring_gpa, no_tutoring_gpa)

# Display the results
t_stat, p_value
```
